## Support Vector Machines
- SVM asks if we can directly find the best separator that maximizes $\gamma$, (creating the largest possible gap between classes), while Perceptron just promises to find you a line that seperates correctly (not necessarily the best line) 

$$(w^*, b^*) \in \arg \max_{w,b} \, \gamma(w, b)$$

subject to 

$$y_i(w^T x_i + b) > 0 \quad \forall i$$

- Take every w, b where the classifier with vector w, and bias b that correctly classifies all the data points. Take the classifier with the highest gamma $\gamma$

## Deriving Point-Boundary Distance

1. 
$$
x = x_p + d
$$
- If xp is the point where the classification line H intersects with the x-axis, then x = xp + d, where d is the perpendicular distance from the point x to the classification line H 
2. 
$$
d = \alpha w \qquad (d \parallel w)
$$
- If w is the vector that is normal to the classification line H, then d is some $\alpha$ scaler of w
3. 
$$
w^\top x_p + b = 0
$$
- Because xp is on the classification line H, we know that w<sup>T</sup>xp + b = 0 by the definition of the classificaiton line 
4. 
$$
0 = w^\top (x - \alpha w) + b
$$
$$
\alpha = \frac{w^\top x + b}{\|w\|_2^2}
$$
- We can substitute in order to find alpha, or the scaling factor 
5. 
$$
\operatorname{dist}(x,H)
= \|d\|_2
= |\alpha|\,\|w\|_2
= \frac{|w^\top x + b|}{\|w\|_2}
$$
- The distance between x and classification line H, is the euclidean distance length of vector d, which is the product of alpha and the euclidean distance length of vector w (the normal vector to classification line H)
6. 
$$
\gamma(w,b)
=
\min_{x \in D_{\mathrm{train}}}
\frac{|w^\top x + b|}{\|w\|_2}
$$
- Gamma is the smallest, or the closest distance from the points in the training set to the classification line, so we iterate through every point in the training data and then find the smallest distance, so that every point is at least $\gamma$ away from the classification line H 

## Hard Margin SVM Optimization

**Idea**: The idea behind hard margin optimization for Support Vector Machines is to fine the separating hyperplane that leaves the largest possible gap between the two classes. 


$$
\max_{w,b} \gamma(w, b) \quad \text{s.t. } y_i(w^T x_i + b) > 0 \, \forall i
$$
- This is the general equation for SVM 

$$
\min_{w,b} \left| w^T x_i + b \right| \quad \max_{w,b} \quad \frac{1}{\|w\|_2} \quad \text{s.t. } y_i(w^T x_i + b) > 0
$$
- Substitute in the derived point boundary distance formula

$$
\text{sign}((\beta w)^T x + \beta b) = \text{sign}(w^T x + b) = \text{sign}(w^T x), \quad \gamma(\beta w, \beta b) = \gamma(w, b)
$$
- Shows that by scaling w and b, we dont actually change the hyperplane

$$
m = \min_i |w^T x_i + b|, \quad \beta = 1/m \quad \Rightarrow \quad \min_{w,b} \|w\|_2^2 \quad \text{s.t. } y_i(w^T x_i + b) \geq 1 \, \forall i
$$
- We choose m to be the smallest absolute functional score, and then set $\beta$ to 1/m so that the closets point's functional score becomes exactly 1
- Because we know that the minimum distance funtional score must be 1, we can substitute it into the $\gamma$ equation so that we get 1/||w<sup>'</sup>||

$$
\max_{w,b} \frac{1}{\|w\|_2} \quad \text{s.t.} \quad y_i(w^\top x_i + b) > 0, \min_i |w^\top x_i + b| = 1
$$

$$
\max \frac{1}{\|w\|_2} \quad \leftrightarrow \quad \min \|w\|_2 \quad \leftrightarrow \quad \min \|w\|_2^2
$$
- Maximizing a fraction, is the same as minimizing its reciprocal

$$
y_i(w^\top x_i + b) > 0, \quad |w^\top x_i + b| \geq 1 \quad \leftrightarrow \quad y_i(w^\top x_i + b) \geq 1
$$
- Minimizing the norm is the same as minimizing its square

$$
\min_{w,b} \|w\|_2^2 \quad \text{s.t.} \quad y_i(w^\top x_i + b) \geq 1 \quad \forall i
$$

- The functional score of a data point at the hard margin optimum will equal 1.
- By scaling w and b, you dont change the classification line H, but we decrease it to 1/m so the support vectors are actually where it hits the first data point, where the functional score is 1, and thats the closest possible points to the classification lines 

## Slack Variables (Soft Margin SVM Optimization) 
- If data is not linearly seperable, then hard-margin constraits are infeasible, and we use slack variables to allow for margin violations
- If a data point has a functional score of 1, then its on the support vector margin, if its inbetween 0 and 1 it's on the correct side but inside the margin, and if its less than 0 its on the wrong side of the margin
$$
\begin{aligned}
& \min_{w,b,\xi} \quad ||w||^2 + C \sum_i \xi_i \\
& \text{s.t.} \quad y_i \left( w^\top x_i + b \right) \geq 1 - \xi_i, \quad \xi_i \geq 0
\end{aligned}
$$
- $\xi$<sub>i</sub> is a slack variable for one data point, measures how much that point is allowed to violate the ideal hard-margin constraint, xi measures how far away the functional score is from 1 
- C controls how much we care about violations, a large C means that violations are expensive, so the model tries to keep $\xi$ very small, whereas a small C means violations are cheaper, so the model is more willing to tolerate points about the margin. C is a hyperparameter that gets chosen by the model developer
- Overall with soft SVM optimiation were trying to make ||w||<sup>2</sup> small in order get a wide margin and make $\xi$ small to avoid margin violations 

Fix \(w, b\). For \(C > 0\), each slack variable should be as small as possible.

$$
\xi_i \geq 0, \qquad \xi_i \geq 1 - y_i(w^\top x_i + b)
$$

$$
\xi_i^* =
\begin{cases}
0, & y_i(w^\top x_i + b) \geq 1, \\
1 - y_i(w^\top x_i + b), & y_i(w^\top x_i + b) < 1
\end{cases}
$$

$$
\xi_i^* = \max\{0, 1 - y_i(w^\top x_i + b)\}
$$

Substitute the minimizing slacks into the objective:

$$
\min_{w,b} \|w\|^2 + C \sum_{i=1}^{n} \max\{0, 1 - y_i(w^\top x_i + b)\}
$$

- THe idea is that for every slack variable (somethign that doesnt fit into classification), the slack variable is the difference between 1 and the functional score, and then we multiply C (our hyperparameter) by the maimum of 0 (if its classified correctly) and our slck variable value, and then were combining the two parts
