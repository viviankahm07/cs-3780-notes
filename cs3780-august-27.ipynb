## ML Fundamentals
- Set inputs as **feature vectors** that someone capture the information through feature extraction - ex. putting a 8x8 image into a 64 length vector
- **Output space/label space**: space of all possible outcomes (ex. digits 1 - 9)
- **Label**: what the data gets classified as (ex. the digit 7)
- Requires a training dataset, that has labelled examples
- **Classifier**: takes an input and outputs its label, maps feature vectors to the label space
- **Classification error/loss function**: tells you the cost of making a mistake
$$
\ell_{0-1}(h(x), y) = \mathbb{1}\{h(x) \neq y\}
$$
- EVERY ML algorithm makes an assumption
- **Main objects**: feature space, labels, labelled examples, and classifier

## K-Nearest Neighbor Classifier
- Classify data based on its nearest neighbor
- Classification makes the assumption that similar inputs are likely to have similar labels 
- Ask the k closest labeled example and return their most common label
1) Measure distance to every point in our training sample
2) Take the k-closest ones
3) The label that comes out as the majority is the answer
- k-nearest neighbor doesn't exapmle have a "model" to train, it's just a set of data to access to get an answer (training time is 0)
- One disadvantage is that we must ensurance that the way we calculate distance has to capture the notion of similarity
- We take the majority through using mode()

## Tie-breaking and Edge Cases 
- Vote tie: if you have a tie between votes between k neighbors, keep decreasing the number of nearest neighbors to k-1 until you eliminate ties
- Distance tie: Use a deterministic rule, such as the first one to show up in the dataset to determine which one is considered the "closest"
- Weighted KNN: Have a distance weighted notion of nearest neighbor, allows closer points to have more influence

## Modelling Distance 
- The classifier can only see the similarities you encode, a good metric makes examples with similar labels close
- You can optimize KNN by coming up with a very accurate model for distance, or having strong feature extraction

## Minkowski Distances
$$
d_p(x,x') = \left(\sum_{j=1}^{d} \left|x_j - x'_j\right|^p\right)^{1/p}
$$

- When p = 1, the Minkowski distance is just the summation of the distances between each coordinate in the vector **(Manhattan distance)**
- When p = 2, the Minkowski distance is the square root of the summation of the square of the distances between each coordinate in the vector **(Euclidean distance)**
- When p goes to $\infty$, the Minkowski distance is the maximum coordinate gap, becuase the maximum element dominates
- When p goes to 0, it counts the number of coordinates where x and x' dont match

## How Does Changing K Affect The Model?
- k = 1: very local, sensitive to noise
- moderate k: local consensus
- k = n: constant classifier, global majority (could be the feature extraction was poor, or there's only one type of classification) 

## Bayes Optimal Predictor 
- The Bayes Optimal Predictor is hypothetical
- The Bayes error is the minimum error you would make even if you had a complete view of the data set
- You can never do better than the Bayes Optimal Predictor
- Hypotehtical because you dont actually know the true underlying distribution
$$
h_{\text{Bayes}}(x) = \arg\max_{y \in \mathcal{Y}} P(Y=y \mid X=x)
$$
$$
\text{Bayes error at } x = 1 - \max_{y \in \mathcal{Y}} P(Y=y \mid X=x)
$$
$$
\text{Bayes classifier expected error}
= \mathbb{E}_X \left[1 - \max_{y \in \mathcal{Y}} P(Y=y \mid X)\right]
$$

- Bayes optimal error predictor observes an event x, and then magically knows the probability that the event takes on every label y. h<sub>Bayes</sub>(x) returns the label that has the highest probability out of all the possibly labels in Y
- The Bayes error at x is the difference between 1 and the Bayes optimal error predictor for the label that has the highest probability
- The Bayes classifier expected error is the expectation of the Bayes error over all possible events x that could occur
- The Bayes Optimal Predictor is hypothetical, because no one knows the true distribution (gold standard) 

## Curse of Dimensionality 
- Dimensions is the number of features in the input vector
- In sufficiently high dimensions, the choice of distance metric no longer matters
- As the number of dimensions increases, points become extremely spread out, so you need exponentially more training data before a "nearest neighbor" is actualy near you

## 1-NN
- 1 Nearest Neighbor can work when you have a lot of examples because you may have a data point that is in the exact location of the input you're trying to predict
- Error is bounded by 2 times the Bayes Optimal Error
