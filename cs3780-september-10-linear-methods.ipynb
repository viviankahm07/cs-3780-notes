- The soft-margin SVM from the previous lecture is already a regularized empirical-risk minimization problem

- The point of the regularizer is to maximize the margin
- Because the training loss (classify the trianing points well) and the second point (maximize the margin) by having a small ||w|| fight each other, and so by minimizing both we try to balance training errors/margin violations against a large margin

## Surrogate Loss 

$$
\underbrace{1\{yf(x) \leq 0\}}_{\ell_{0-1}(y,f(x))}
\leq
\underbrace{\max\{0,1-yf(x)\}}_{\ell_{\text{hinge}}(y,f(x))}
$$

- Hinge loss is a surrogate loss
- Becuase a classification error (0-1) loss can be optimize, we optimize hinge loss as a conveninent substitute

## Logistic Regression Objective 

$$
\min_{w,b} \frac{1}{n} \sum_{i=1}^{n} \log \left( 1 + e^{-y_i(w^\top x_i + b)} \right) + \frac{\lambda}{2} \|w\|_2^2
$$

**Logistic Loss:**
$$
\frac{1}{n} \sum_{i=1}^{n} \log \left( 1 + e^{-y_i(w^\top x_i + b)} \right)
$$

- Assume that input, output pairs are drawn iid from some kind of fixed distribution
- Logistic regression makes the assumption that...

$$
P(Y = 1 \mid X = x) = \frac{1}{1 + \exp(- (w^\top x + b))} = \sigma(w^\top x + b)
$$
- Uses a sigmoid function
- Dictates that as you move from 0 to 1 (representing +1 and -1), the larger the value the score is the higher the probability is that the label is 1 

## A Probabilitic Interpretation
$$(\hat{w}^{MLE}, \hat{b}^{MLE}) = \arg \max_{w, b} P(D \mid w, b)$$
- Definition of the maximum likelihood estimator, MLE for w and b that maximizes the likelihood of data
$$
= \arg \max_{w,b} P((y_1, x_1), \ldots, (y_n, x_n) | w, b)
$$
- Substituting in the individual probabilities
$$
\text{arg max}_{w,b} \prod_{i=1}^{n} P(y_i, x_i | w, b)
$$
- Because we assume that probabilities are generated iid
$$
\arg \max_{w,b} \prod_{i=1}^{n} P(y_i \mid x_i, w, b) P(x_i \mid w, b)
$$
- Breaking it down into conditional probabilities (so that you get the probability of y being label y- given the w and b were optimizing ```markdown
$$
\arg \max_{w,b} \prod_{i=1}^{n} P(y_i \mid x_i, w, b) P(x_i)
$$
- Simplify becuase xi is not dependent on w and b
$$
\text{arg max}_{w,b} \prod_{i=1}^{n} P(y_i \mid x_i, w, b)
$$
- You can drop the P(xi) because its a constant/single factor 
$$ 
= \arg \max_{w,b} \sum_{i=1}^n \log P(y_i \mid x_i, w, b) 
$$
- Instead of doing a product of probabilities, you can take the log so it becomes a sum of probabilities. When probability are small multiplying probabilities becomes even smaller so adding is better
$$
\begin{align*}
\text{arg max}_{w,b} & \quad -\sum_{i=1}^{n} \log \left( 1 + e^{-y_i (w^T x_i + b)} \right) \\
\\
\text{arg min}_{w,b} & \quad \sum_{i=1}^{n} \log \left( 1 + e^{-y_i (w^T x_i + b)} \right)
\end{align*}
$$

## Recap of Soft Margin Support Vector Machine 


1. **Wide Margin, Penalize Total Slack, Allow for Violations**

   $$ 
   \min_{w,b,\xi} \quad \frac{1}{2} \|w\|_2^2 + C \sum_{i=1}^n \xi_i \quad \text{s.t.} \quad y_i (w^\top x_i + b) \geq 1 - \xi_i, \quad \xi_i \geq 0 
   $$

- For every possible weight vector w, bias b, and slack variable $\xi$, we want to minimize this equation
- 1/2||w||<sup>2</sup> encouragea wide margin without accounting for slack variables 
- We create a slack variable $\xi$ that determines the amoutn of margin violation we are allowing for point i
- We then penalize large $\xi$'s through the summation
- C tells the SVM how ecpensive slack/violations are -> large C punishes heavily, a small C tolerates violations
- We dont choose the error $\xi$, its the number needed to make the y<sub>i</sub> (w<sup>T</sup>*x<sub>i</sub> + b) >= 1 - $\xi$ true 
---

2. **Equivalent Formulation**
   
   $$ 
   \min_{w,b} \quad \frac{1}{2} \|w\|_2^2 + C \sum_{i=1}^n \max\{0, 1 - y_i (w^\top x_i + b)\}
   $$

---

3. **Normalized Notation**

   $$
   \min_{w,b} \quad \sum_{i=1}^n \ell_{\text{hinge}}(h(x_i), y_i) + \lambda \|w\|_2^2
   $$

   Where \( \lambda = \frac{1}{2C} \)
- Then we can just write it in terms of the hinge loss equation
