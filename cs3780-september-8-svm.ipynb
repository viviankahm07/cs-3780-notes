## Support Vector Machines
- SVM asks if we can directly find the best separator that maximizes $\gamma$, (creating the largest possible gap between classes), while Perceptron just promises to find you a line that seperates correctly (not necessarily the best line) 

$$(w^*, b^*) \in \arg \max_{w,b} \, \gamma(w, b)$$

subject to 

$$y_i(w^T x_i + b) > 0 \quad \forall i$$

- Take every w, b where the classifier with vector w, and bias b that correctly classifies all the data points. Take the classifier with the highest gamma $\gamma$

## Deriving Point-Boundary Distance

1. 
$$
x = x_p + d
$$
- If xp is the point where the classification line H intersects with the x-axis, then x = xp + d, where d is the perpendicular distance from the point x to the classification line H 
2. 
$$
d = \alpha w \qquad (d \parallel w)
$$
- If w is the vector that is normal to the classification line H, then d is some $\alpha$ scaler of w
3. 
$$
w^\top x_p + b = 0
$$
- Because xp is on the classification line H, we know that w<sup>T</sup>xp + b = 0 by the definition of the classificaiton line 
4. 
$$
0 = w^\top (x - \alpha w) + b
$$
$$
\alpha = \frac{w^\top x + b}{\|w\|_2^2}
$$
- We can substitute in order to find alpha, or the scaling factor 
5. 
$$
\operatorname{dist}(x,H)
= \|d\|_2
= |\alpha|\,\|w\|_2
= \frac{|w^\top x + b|}{\|w\|_2}
$$
- The distance between x and classification line H, is the euclidean distance length of vector d, which is the product of alpha and the euclidean distance length of vector w (the normal vector to classification line H)
6. 
$$
\gamma(w,b)
=
\min_{x \in D_{\mathrm{train}}}
\frac{|w^\top x + b|}{\|w\|_2}
$$
- Gamma is the smallest, or the closest distance from the points in the training set to the classification line, so we iterate through every point in the training data and then find the smallest distance, so that every point is at least $\gamma$ away from the classification line H 

## Hard Margin SVM Optimization

**Idea**: The idea behind hard margin optimization for Support Vector Machines is to fine the separating hyperplane that leaves the largest possible gap between the two classes. 


$$
\max_{w,b} \gamma(w, b) \quad \text{s.t. } y_i(w^T x_i + b) > 0 \, \forall i
$$
- This is the general equation for SVM 

$$
\min_{w,b} \left| w^T x_i + b \right| \quad \max_{w,b} \quad \frac{1}{\|w\|_2} \quad \text{s.t. } y_i(w^T x_i + b) > 0
$$
- Substitute in the derived point boundary distance formula

$$
\text{sign}((\beta w)^T x + \beta b) = \text{sign}(w^T x + b) = \text{sign}(w^T x), \quad \gamma(\beta w, \beta b) = \gamma(w, b)
$$
- Shows that by scaling w and b, we dont actually change the hyperplane

$$
m = \min_i |w^T x_i + b|, \quad \beta = 1/m \quad \Rightarrow \quad \min_{w,b} \|w\|_2^2 \quad \text{s.t. } y_i(w^T x_i + b) \geq 1 \, \forall i
$$

$$
\max_{w,b} \frac{1}{\|w\|_2} \quad \text{s.t.} \quad y_i(w^\top x_i + b) > 0, \min_i |w^\top x_i + b| = 1
$$


$$
\max \frac{1}{\|w\|_2} \quad \leftrightarrow \quad \min \|w\|_2 \quad \leftrightarrow \quad \min \|w\|_2^2
$$


$$
y_i(w^\top x_i + b) > 0, \quad |w^\top x_i + b| \geq 1 \quad \leftrightarrow \quad y_i(w^\top x_i + b) \geq 1
$$


$$
\min_{w,b} \|w\|_2^2 \quad \text{s.t.} \quad y_i(w^\top x_i + b) \geq 1 \quad \forall i
