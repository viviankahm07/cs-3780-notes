## Linear Methods - Regression

### Recap of Regularized Empirical Risk Minimization
$$
\hat{h} \in \arg \min_{h \in \mathcal{H}} \left( \frac{1}{n} \sum_{i=1}^{n} \ell(h(x_i), y_i) + \lambda R(h) \right)
$$

- We take training samples 1 through n, and we want to minimize the loss + some term that is the regularizer
- The tradeoff is to make the data accurate to the training data, while the regularizer expresses a preference among predictors, like reducing complexity to prevent overfitting

### MLE Becomes Least Squares (Ordinary Least Squares) 

$$(\hat{w}^{MLE}, \hat{b}^{MLE}) = \arg \max_{w,b} P(\mathcal{D} | w, b)$$
- We are trying to choose the w and b that create the maximum probability of seeing the training data from w and b

$$= \arg \max_{w,b} P((y_1, x_1), \ldots, (y_n, x_n) | w, b)$$
- Because these are independent distributions, we can split up and multiply the probabilitie s

$$= \arg \max_{w,b} \prod_{i=1}^{n} P(y_i | x_i | w, b)$$
- Use a summation formula 

$$= \arg \max_{w,b} \prod_{i=1}^{n} P(y_i | x_i, w, b) P(x_i)$$
- Split up the into conditional probabilities and condition on each other 

$$= \arg \max_{w,b} \prod_{i=1}^{n} P(y_i | x_i, w, b)$$
- use log to turn the product of probabilities into a summation, the w and b that optimize the product is the same as the w and b that optimize the log summation 

$$= \arg \max_{w,b} \sum_{i=1}^{n} \log P(y_i | x_i, w, b)$$
- Substituting in the Gaussian model because we assume that it follows a normal distribution 

$$= \arg \max_{w,b} \sum_{i=1}^{n} \log \frac{1}{\sqrt{2\pi \sigma^2}} e^{-\frac{(y_i - w^\top x_i - b)^2}{2\sigma^2}}$$

$$= \arg \max_{w,b} -\frac{1}{2\sigma^2} \sum_{i=1}^{n} (y_i - w^\top x_i - b)^2 + \text{const}$$
- Because the log and the exponential cancel out 

$$= \arg \min_{w,b} \frac{1}{n} \sum_{i=1}^{n} (y_i - w^\top x_i)^2$$
- We always prefer to minimize instead of maximization, so convert 

### Maximum A Posteriori Gives Ridge Regression 
- I have some prior belief on what w is
- And maybe I only have a small amount of data, so I can put some kind of prior on the estimate in combination with the data that I actually have
- Were trying to maximize the w and b that have the max posterior probability based on the prior probability
- We're going to do a simple method to finding the MLE but were going to plug in our prior estimate probability
$$
w \sim \mathcal{N}(0, \tau^2 I), \quad P(w) = \frac{1}{(2\pi\tau^2)^{d/2}} \exp\left(-\frac{\|w\|_2^2}{2\tau^2}\right)
$$

$$(\hat{w}_{MAP}, \hat{b}_{MAP}) = \arg \max_{w,b} P(w, b \mid D) = \arg \max_{w,b} P(D \mid w, b)P(w)$$

$$
= \arg \max_{w,b} \left( \prod_{i=1}^{n} P(y_i \mid x_i, w, b) \right) P(w)
$$

$$
= \arg \max_{w,b} \sum_{i=1}^{n} \log P(y_i \mid x_i, w, b) + \log P(w)
$$

$$
= \arg \max_{w,b} -\sum_{i=1}^{n} \log P(y_i \mid x_i, w, b) - \log P(w)
$$

$$
= \arg \min_{w,b} \frac{1}{2\sigma^2} \sum_{i=1}^{n} (y_i - w^\top x_i - b)^2 + \frac{1}{2\tau^2} \|w\|_2^2
$$

$$
= \arg \min_{w,b} \frac{1}{n} \sum_{i=1}^{n} (y_i - w^\top x_i - b)^2 + \lambda \|w\|_2^2, \quad \lambda = \frac{\sigma^2}{n\tau^2}
$$

$$
\hat{w}_{ridge} = \arg \min_{w,b} \frac{1}{n} \sum_{i=1}^{n} (y_i - w^\top x_i - b)^2 + \lambda \|w\|_2^2
$$

- If our data has a large variance, then our $\lambda$ becomes larger and larger, in the other direction, it dissapears to 0
- Find w and b that minimize the sum of squared loss + $\lambda$ * norm of w<sup>2</sup> which is essentially the regularizer

### Closed Form Solution
- A closed form solution means you can write down the answer as a finite formula instead of finding it iteratively
#### OLS Objective
$$
\hat{w}_{OLS} = \arg \min_{w} \frac{1}{n} \sum_{i=1}^{n} (x_i^\top w - y_i)^2
$$

#### OLS Closed Form
$$
\hat{w}_{OLS} = (X^\top X)^{-1} X y
$$

#### Ridge Regression Objective
$$
\hat{w}_{ridge} = \arg \min_{w} \frac{1}{n} \sum_{i=1}^{n} (x_i^\top w - y_i)^2 + \lambda \|w\|_2^2
$$

#### Ridge RegressionClosed Form
$$
\hat{w}_{ridge} = (X^\top X + n \lambda I)^{-1} X y
$$
- X is a matrix with d rows and n columns, where each column is one distance of the data x<sub>i</sub>

### Regularizer's Choice 
- What are the options for a regularizer in terms of a linear classifier
- The lambda tells us how to balance the two parts of the regularized ERM 
- l<sub>2</sub> (ridge): $$
r(w) = \|w\|^2_2 = w^\top w
$$
- For l2, you usually get a circle 
- l<sub>1</sub> (lasso): $$
r(w) = \|w\|_1 = \sum_{j=1}^{d} |w_j|
$$
- For l1, you usually get a square
