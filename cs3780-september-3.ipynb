## Improving a Linear Classifier 
- If we misclassify a data point, we move away from the misclassified negative points
- For an incorrect negative example, take the currnet weight **w**, and subtract off the misclassified point **x**
- For an incorrect positive example, take the current weight **w**, and add the miscalssified point **x**
$$
\begin{cases}
\vec{w} + \vec{x}_i & y_i = 1 \\
\vec{w} - \vec{x}_i & y_i = -1
\end{cases}
\quad \Longleftrightarrow \quad
\vec{w} + y_i\vec{x}_i
$$

## The Perception Algorithm 
- **Input:** training data $S = \{(\vec{x}_1, y_1), \ldots, (\vec{x}_m, y_m)\}$
- **Initialize** $\vec{w}^{(0)} = (0, \ldots, 0)$ and $t = 0$
- **While** there is $i \in [m]$ such that $y_i(\vec{w}^{(t)} \cdot \vec{x}_i) \leq 0$ then,
  - $\vec{w}^{(t+1)} = \vec{w}^{(t)} + y_i\vec{x}_i$
  - $t \leftarrow t + 1$
- **End While**
- **Output** $\vec{w}^{(t)}$

## Perceptron Example
- x1 = (1, 2), x2 = (2, 1), x3 = (-1, -1), x4 = (-1, 1)
- y1 = 1, y2 = 1, y3 = -1, y4= -1
- Start with a weight vector w<sup>0</sup> of (0, 0)
- Pick data point x1, and update our weight vector w<sup>1</sup> to (1, 2)
- Data point 1 is now correctly classified (you can use the equation to check that y<sub>1</sub>*(w<sub>1</sub>x<sub>1</sub>) is greater than 0)
- Data point 2 and 3 are already correctly classified based on w<sub>1</sub> since the dot product is > 0
- Data point 4 is incorrectly classified because the dot product is less than 0, so we must update that w<sup>2</sup> = w<sup>1</sup> - x<sub>4</sub> = (2, 1), and then recheck that all the other data points are still correctly classified

- There is no obvious choice about which point in the dataset is the best to start from 

## Margin 
- A data set S is linearly seperaable with a **geometric) margin $\gamma$** if
- 1) There is a linear classifier h<sub>w</sub> that is consistent with S
  2) The distance of any instance S to the decision boundary of H<sub>w</sub> is at least $\gamma$
- Mathematical definition: There is $\vec{w}$ such that $\|\vec{w}\| = 1$ and for all data points $(\vec{x}_i, y_i) \in S$

  $\begin{cases}
  \vec{w} \cdot \vec{x}_i \geq \gamma & y_i = 1 \\
  \vec{w} \cdot \vec{x}_i \leq -\gamma & y_i = -1
  \end{cases}
  \Longleftrightarrow
  y_i(\vec{w} \cdot \vec{x}_i) \geq \gamma$

## Convergence of Perceptron 
- **Theorem:** Given a dataset $S = \{(\vec{x}_1, y_1), \ldots, (\vec{x}_m, y_m)\}$ and a radius $R$ such that $\|\vec{x}_i\| \leq R$ for all $i \in [m]$. If $S$ is linearly separable with (geometric) **margin** $\gamma$, then Perceptron makes at most $R^2/\gamma^2$ updates before finding a consistent linear classifier.

## Proof Outline
- Goal: Prove that Perceptron terminates after at most $R^2 / \gamma^2$ steps
- Step 1: Show that $\vec{w}^{*} \cdot \vec{w}^{(t)}$ is large (find a lower bound)
- Step 2: Show that $\|\vec{w}^{(t)}\|$ is not too large (find an upper bound)
- Step 3: Combine 1 & 2 to show that if there are more than $R^2 / \gamma^2$ updates, then the cosine becomes larger than one!
  - Proof by contradiction

## Step 1: $\vec{w}^{*} \cdot \vec{w}^{(t)}$ is large

- Goal: lower bound $\vec{w}^{*} \cdot \vec{w}^{(t)}$ in terms of $\gamma$ and $t$

- First, bound $\vec{w}^{*} \cdot \vec{w}^{(t)}$ in terms of $\vec{w}^{*} \cdot \vec{w}^{(t-1)}$ and $\gamma$
  - Let $(\vec{x}_i, y_i)$ be the mistake made at iteration $t$
  - $\vec{w}^{*} \cdot \vec{w}^{(t)} = \vec{w}^{*} \cdot \left(\vec{w}^{(t-1)} + y_i\vec{x}_i\right)$
    - *Definition of perceptron update*
  - $= \vec{w}^{*} \cdot \vec{w}^{(t-1)} + y_i\vec{w}^{*} \cdot \vec{x}_i$
    - *Distributive property of dot products, commutative scalars*
  - $\geq \vec{w}^{*} \cdot \vec{w}^{(t-1)} + \gamma$
    - *Definition of $\vec{w}^{*}$ and margin*
- Then, recurse to bound $\vec{w}^{*} \cdot \vec{w}^{(t)}$ in terms of $\gamma$ and $t$
  - $\vec{w}^{*} \cdot \vec{w}^{(t)} \geq \vec{w}^{*} \cdot \vec{w}^{(t-2)} + 2\gamma$
  - $\geq \vec{w}^{*} \cdot \vec{w}^{(t-3)} + 3\gamma$
  - $\cdots$
  - $\geq \vec{w}^{*} \cdot \vec{w}^{(0)} + t\gamma$

## Step 2: $\|\vec{w}^{(t)}\|$ is not too large

- Goal: upper bound $\|\vec{w}^{(t)}\|$ in terms of $R$ and $t$

- First, bound $\|\vec{w}^{(t)}\|^2$ in terms of $\|\vec{w}^{(t-1)}\|^2$ and $R$

  Let $(\vec{x}_i, y_i)$ be the mistake made at iteration $t$

  1. $\|\vec{w}^{(t)}\|^2 = \|\vec{w}^{(t-1)} + y_i\vec{x}_i\|^2$
     - *Definition of perceptron update*

  2. $= (\vec{w}^{(t-1)} + y_i\vec{x}_i) \cdot (\vec{w}^{(t-1)} + y_i\vec{x}_i)$
     - *Definition of norm/dot product*

  3. $= \vec{w}^{(t-1)} \cdot \vec{w}^{(t-1)} + 2y_i(\vec{w}^{(t-1)} \cdot \vec{x}_i) + (y_i)^2\vec{x}_i \cdot \vec{x}_i$
     - *Multiplying and combining terms*

  4. $\leq \|\vec{w}^{(t-1)}\|^2 + 0 + R^2$
     - *Mistake + radius assumption*

- Therefore,

  $\|\vec{w}^{(t)}\|^2 \leq \|\vec{w}^{(t-1)}\|^2 + R^2$

- Then, recurse to bound $\|\vec{w}^{(t)}\|^2$ in terms of $R$ and $t$

  - $\|\vec{w}^{(t)}\|^2 \leq \|\vec{w}^{(t-2)}\|^2 + 2R^2$
  - $\cdots$
  - $\|\vec{w}^{(t)}\|^2 \leq \|\vec{w}^{(0)}\|^2 + tR^2$

## Step 3: Contradiction

- Goal: show that if $t > R^2/\gamma^2$, there is a contradiction
- We will use Step 1 $(\vec{w}^{*} \cdot \vec{w}^{(t)} \geq t\gamma)$ & Step 2 $(\|\vec{w}^{(t)}\|^2 \leq tR^2)$
- Starting with our expression for cosine,
  - $\cos(\theta_{\vec{w}^{*},\vec{w}^{(t)}}) = \frac{\vec{w}^{*} \cdot \vec{w}^{(t)}}{\|\vec{w}^{*}\|\|\vec{w}^{(t)}\|}$
  - $= \frac{\vec{w}^{*} \cdot \vec{w}^{(t)}}{\|\vec{w}^{(t)}\|}$
    - *Definition of $\vec{w}^{*}$*
  - $\geq \frac{t\gamma}{\|\vec{w}^{(t)}\|}$
    - *Step 1*
  - $\geq \frac{t\gamma}{\sqrt{tR^2}}$
    - *Step 2*
  - $= \frac{\sqrt{t}\gamma}{R}$
    - *Division*

- If $t > R^2/\gamma^2$,
  - $\cos(\theta_{\vec{w}^{*},\vec{w}^{(t)}}) > \frac{\sqrt{R^2/\gamma^2}\gamma}{R} = 1 \Longrightarrow$ contradiction!

## Remarks on Convergence 
- The actual number of updaes can vary depending on scaling, shuffling the training set, or the value of $\gamma$, but we know it cannot be more than R<sup>2</sup>/$\gamma$<sup>2</sup>
