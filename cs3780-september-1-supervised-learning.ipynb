## Vector Review 
- **Euclidean norm**
$$
\|\vec{x}\|
=
\sqrt{x_1^2 + x_2^2 + \cdots + x_n^2}
=
\sqrt{\sum_{i=1}^{n} x_i^2}
$$
- **Dot product:**

$$
\vec{w} \cdot \vec{x}
=
w_1x_1 + w_2x_2 + \cdots + w_nx_n
=
\sum_{i=1}^{n} w_i x_i
$$
- The dot product is the product of the norms of the two vectors * cosine of the angle between them
- Projection of one vector onto another vector
- If the angle between 2 vectors is acute, then the dot product > 0
- If the angel between 2 vectors is 90, then the dot product = 0
- If the angle between 2 vectors is obtuse, then the dot product < 0

## Hyperplanes 
- In d dimensional space, a hyperplane is defined by a vector w in R<sup>d</sup> and a scalar b in R
$$
\left\{ \vec{x} \in \mathbb{R}^d \mid \vec{w} \cdot \vec{x} + b = 0 \right\}
$$
- b is the shift of the hyperplane away from the origin
- The hyperplane is orthogonal to w
- The distance to the origin is -b/||w||
- All points on the hyperplane have w * x = -b

## Linear Classifier

- **Linear Classifier:**
$$
h_{\vec{w}, b} : \mathbb{R}^d \rightarrow \{-1, +1\}
$$

- Has a **weight vector** $\vec{w}$ and a **bias** $b$.
- Takes an input $\vec{x}$ and classifies it based on which side of the decision boundary it falls on:

$$
h_{\vec{w}, b}(\vec{x})
=
\operatorname{sign}(\vec{w}\cdot\vec{x}+b)
=
\begin{cases}
+1 & \vec{w}\cdot\vec{x}+b>0 \\
-1 & \vec{w}\cdot\vec{x}+b\le 0
\end{cases}
$$

- The decision boundary is defined by:
$$
\vec{w}\cdot\vec{x}+b=0
$$

- In **1 dimension**, the decision boundary is a point.
- In **2 dimensions**, the decision boundary is a line.
- In **3 dimensions**, the decision boundary is a plane.
- In **$d$ dimensions**, the decision boundary is a $(d-1)$-dimensional **hyperplane**.

## Homogenous Linear Classifiers
- A classifier is homogenous if b = 0 (runs through the origin), otherwise it is non homogenous
- Any d dimensional learning problem for linear classifiers has a homogenous form in d + 1 dimensions

## Supervised Learning
- **Features**: x in X
- **Labels**: y in Y
- **Data**: D = {(x<sub>1</sub>, y<sub>1</sub>), ...(x<sub>n</sub>, y<sub>n</sub>)
- **Model Class**: H = {h: X -> Y}
- **Loss Function**: l: Y x Y -> R (real number)
- Learning algorithm that picks an h in H based on D and l

## Loss Functions

- There are different kinds of loss functions for different kinds of labels
- **Classification Loss:**

$$
\ell(y, y')
=
\begin{cases}
1 & \text{if } y \ne y' \\
0 & \text{otherwise } (y = y')
\end{cases}
$$

- **Square Loss:**

$$
\ell(y, y') = \|y - y'\|^2
$$

- **Absolute Loss:**

$$
\ell(y, y') = |y - y'|
$$

## Generalization 
- Hypothesis that genearlize well to new data 

## Train/Test Split
- Split the data into two parts, train and test, D = D<sub>train</sub> U D<sub>test</sub>
- Choose a hypothesis based on D<sub>train</sub>
- Evaluate the hypothesis on D<sub>test</sub>
$$
\frac{1}{n}
\sum_{i=1}^{n}
\ell\bigl(y_i, h(x_i)\bigr)
\quad
\text{for } (x_i, y_i) \in \mathcal{D}_{\text{test}}
$$
- Measures how well our model generalized to new data that we havent trained on
